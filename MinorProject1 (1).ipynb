Project Title

Student Marks Analyzer using Python and NumPy

Problem Statement

The aim of this project is to create a Python program that accepts marks of multiple students, stores them using NumPy arrays, and performs basic analysis such as total marks, average marks, highest and lowest marks. Based on the average marks, the program assigns grades using conditional statements.

Objectives

To understand usage of Python basics

To work with NumPy arrays

To apply loops, conditions, and functions

To analyze numerical data logically


Tools & Concepts Used

Python basics

input() and print()

Lists

Loops (for)

Conditional statements (if-elif-else)

User-defined functions

NumPy array operations (sum, mean, max, min)

```python
# Student Marks Analyzer
# Import NumPy library
import numpy as np

# Function to assign grade based on average marks
def assign_grade(avg):
    if avg >= 85:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "Fail"

# Taking number of students
n = int(input("Enter number of students: "))

# List to store marks
marks_list = []

# Taking marks input
for i in range(n):
    marks = float(input(f"Enter marks of student {i+1}: "))
    marks_list.append(marks)

# Convert list to NumPy array
marks_array = np.array(marks_list)

# Calculations
total_marks = np.sum(marks_array)
average_marks = np.mean(marks_array)
highest_marks = np.max(marks_array)
lowest_marks = np.min(marks_array)

# Assign grade
grade = assign_grade(average_marks)

# Display results
print("\n--- Student Marks Analysis ---")
print("Marks Entered:", marks_array)
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
print("Grade:", grade)
```
